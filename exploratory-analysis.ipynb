```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

speed_col = "Speed (m/s)"
lat_col = "Lat"
lon_col = "Lon"

speed_threshold = 3.0

for i, df in enumerate(dfs):
    tmp = df.copy()

    tmp[lat_col] = tmp[lat_col].astype(str).str.replace(",", ".", regex=False)
    tmp[lon_col] = tmp[lon_col].astype(str).str.replace(",", ".", regex=False)
    tmp[speed_col] = tmp[speed_col].astype(str).str.replace(",", ".", regex=False)

    tmp[lat_col] = pd.to_numeric(tmp[lat_col], errors="coerce")
    tmp[lon_col] = pd.to_numeric(tmp[lon_col], errors="coerce")
    tmp[speed_col] = pd.to_numeric(tmp[speed_col], errors="coerce")

    tmp = tmp.dropna(subset=[lat_col, lon_col, speed_col]).copy()

    if len(tmp) < 2:
        print(f"dfs[{i}] -> za mało danych")
        continue

    fast = tmp[tmp[speed_col] > speed_threshold]
    if len(fast) < 10:
        fast = tmp

    lat = tmp[lat_col].values
    lon = tmp[lon_col].values
    speed = tmp[speed_col].values

    lat0 = np.mean(fast[lat_col].values)
    lon0 = np.mean(fast[lon_col].values)

    x = lon - lon0
    y = lat - lat0

    mask = speed > speed_threshold
    if mask.sum() < 10:
        mask = np.ones_like(speed, dtype=bool)

    xmin, xmax = np.min(x[mask]), np.max(x[mask])
    ymin, ymax = np.min(y[mask]), np.max(y[mask])

    x_shift = 0.0
    y_shift = 0.0

    if xmin < -0.0005:
        x_shift += -0.0005 - xmin
    if xmax + x_shift > 0.0005:
        x_shift += 0.0005 - (xmax + x_shift)

    if ymin < -0.0003:
        y_shift += -0.0003 - ymin
    if ymax + y_shift > 0.0003:
        y_shift += 0.0003 - (ymax + y_shift)

    x = x + x_shift
    y = y + y_shift

    print(f"dfs[{i}] Lon range: {x.min():.6f} do {x.max():.6f}")
    print(f"dfs[{i}] Lat range: {y.min():.6f} do {y.max():.6f}")

    plt.figure(figsize=(12, 8))

    plt.plot(x, y, linewidth=1.5, color="blue")
    plt.scatter(x[0], y[0], s=50, color="green", label="start")
    plt.scatter(x[-1], y[-1], s=50, color="red", label="end")

    plt.xlabel("Lon (shifted)")
    plt.ylabel("Lat (shifted)")
    plt.title(f"Trajectory of player no {i+1}")

    plt.legend()
    plt.axis("equal")
    plt.grid(True, alpha=0.3)

    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

speed_col = "Speed (m/s)"
lat_col = "Lat"
lon_col = "Lon"

speed_threshold = 3.0

for i, df in enumerate(dfs):
    tmp = df.copy()

    tmp[lat_col] = tmp[lat_col].astype(str).str.replace(",", ".", regex=False)
    tmp[lon_col] = tmp[lon_col].astype(str).str.replace(",", ".", regex=False)
    tmp[speed_col] = tmp[speed_col].astype(str).str.replace(",", ".", regex=False)

    tmp[lat_col] = pd.to_numeric(tmp[lat_col], errors="coerce")
    tmp[lon_col] = pd.to_numeric(tmp[lon_col], errors="coerce")
    tmp[speed_col] = pd.to_numeric(tmp[speed_col], errors="coerce")

    tmp = tmp.dropna(subset=[lat_col, lon_col, speed_col]).copy()

    if len(tmp) < 2:
        print(f"dfs[{i}] -> za mało danych")
        continue

    fast = tmp[tmp[speed_col] > speed_threshold]
    if len(fast) < 10:
        fast = tmp

    lat_all = np.radians(tmp[lat_col].values)
    lon_all = np.radians(tmp[lon_col].values)
    speed_all = tmp[speed_col].values

    lat_fast = np.radians(fast[lat_col].values)
    lon_fast = np.radians(fast[lon_col].values)

    lat0 = np.mean(lat_fast)
    lon0 = np.mean(lon_fast)

    R = 6371000.0

    x = (lon_all - lon0) * np.cos(lat0) * R
    y = (lat_all - lat0) * R

    mask = speed_all > speed_threshold
    if mask.sum() < 10:
        mask = np.ones_like(speed_all, dtype=bool)

    xmin, xmax = np.min(x[mask]), np.max(x[mask])
    ymin, ymax = np.min(y[mask]), np.max(y[mask])

    x_shift = 0.0
    y_shift = 0.0

    if xmin < -52.5:
        x_shift += -52.5 - xmin
    if xmax + x_shift > 52.5:
        x_shift += 52.5 - (xmax + x_shift)

    if ymin < -34:
        y_shift += -34 - ymin
    if ymax + y_shift > 34:
        y_shift += 34 - (ymax + y_shift)

    x = x + x_shift
    y = y + y_shift

    print(f"dfs[{i}] X range: {x.min():.2f} do {x.max():.2f}")
    print(f"dfs[{i}] Y range: {y.min():.2f} do {y.max():.2f}")

    plt.figure(figsize=(12, 8))

    ax = plt.gca()
    ax.add_patch(plt.Rectangle((-52.5, -34), 105, 68, fill=False, linewidth=2))
    plt.plot([0, 0], [-34, 34], linewidth=1.5)
    circle = plt.Circle((0, 0), 9.15, fill=False)
    ax.add_patch(circle)

    ax.add_patch(plt.Rectangle((-52.5, -20.16), 16.5, 40.32, fill=False, linewidth=1.5))
    ax.add_patch(plt.Rectangle((52.5 - 16.5, -20.16), 16.5, 40.32, fill=False, linewidth=1.5))

    plt.plot(x, y, linewidth=1.5, color="blue")
    plt.scatter(x[0], y[0], s=50, color="green")
    plt.scatter(x[-1], y[-1], s=50, color="red")

    plt.xlabel("X [m]")
    plt.ylabel("Y [m]")
    plt.title(f"Trajectory of player no {i+1}")

    plt.axis("equal")
    plt.grid(True, alpha=0.3)

    plt.xlim(-60, 60)
    plt.ylim(-40, 40)

    plt.show()
```

```python
sigma = 0.5  # tolerance under maximum in m/s^2

speed_treshold = 3
selected_speeds, selected_accels = [], []

while speed_treshold <= positive_df["Speed median mean"].max():
    bin_data = positive_df[
        (positive_df["Speed median mean"] >= speed_treshold) &
        (positive_df["Speed median mean"] < speed_treshold + 0.2)
    ]
    if not bin_data.empty:
        selected_speeds.append(bin_data["Speed median mean"].mean())
        selected_accels.append(bin_data["dv/dt median mean"].max() - sigma)
    speed_treshold += 0.2

model = LinearRegression()
model.fit(np.array(selected_speeds).reshape(-1, 1), np.array(selected_accels))
a, b = model.coef_[0], model.intercept_
A0, S0 = b, -b / a

plt.figure(figsize=(12, 8))
plt.scatter(positive_df["Speed median mean"], positive_df["dv/dt median mean"], color="grey", s=5, alpha=0.3)
plt.scatter(selected_speeds, selected_accels, color="red", s=10, alpha=0.8, label="max - sigma per bin")
plt.plot([0, S0], [A0, 0], color="red", linestyle="-", label=f"SFA (sigma={sigma}), A0={A0:.2f}, S0={S0:.2f}")
plt.xlabel("Speed [m/s]")
plt.ylabel("Acceleration [m/s²]")
plt.title(f"ASP - SFA (sigma={sigma})")
plt.legend()
plt.show()
print(f"A0={A0:.3f}, S0={S0:.3f}, nachylenie={a:.4f}")
```
